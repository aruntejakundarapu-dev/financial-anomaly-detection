```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a sample financial dataset
data = {
    'Transaction_ID': range(1, 11),
    'Amount': [
        500, 700, 650, 800, 550,
        600, 750, 680, 5000, 620
    ]
}

df = pd.DataFrame(data)

# Display the dataset
print("Financial Dataset:")
print(df)

# Calculate quartiles
Q1 = df['Amount'].quantile(0.25)
Q3 = df['Amount'].quantile(0.75)

# Calculate Interquartile Range
IQR = Q3 - Q1

# Define lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify anomalies
df['Anomaly'] = (
    (df['Amount'] < lower_limit) |
    (df['Amount'] > upper_limit)
)

print("\nAnomaly Detection Results:")
print(df)

# Visualize transaction amounts
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Transaction_ID',
    y='Amount',
    hue='Anomaly',
    palette={False: 'blue', True: 'red'},
    s=100
)

plt.title('Financial Transaction Anomaly Detection')
plt.xlabel('Transaction ID')
plt.ylabel('Transaction Amount')
plt.grid(True)
plt.show()

# Create a box plot
plt.figure(figsize=(7, 4))

sns.boxplot(x=df['Amount'])
plt.title('Distribution of Financial Transactions')
plt.show()
```
